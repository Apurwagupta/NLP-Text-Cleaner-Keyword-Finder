```python

import re
import nltk
import pandas as pd
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

# Download the English stop-word list
nltk.download("stopwords")

# Load the CSV file
df = pd.read_csv("sample_texts.csv")

# Initialize NLP tools
stop_words = set(stopwords.words("english"))
stemmer = PorterStemmer()

def clean_text(text):
    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)

    # Remove punctuation, numbers and special characters
    text = re.sub(r"[^a-z\s]", " ", text)

    # Split text into individual words
    words = text.split()

    # Remove stop words and apply stemming
    words = [
        stemmer.stem(word)
        for word in words
        if word not in stop_words
    ]

    return " ".join(words)

# Apply text cleaning to every row
df["cleaned_text"] = df["text"].apply(clean_text)

print("Text cleaning completed!")
display(df[["text", "cleaned_text"]].head(10))
```

```python

from collections import Counter
import matplotlib.pyplot as plt

# Combine all cleaned texts into one list of words
all_words = " ".join(df["cleaned_text"]).split()

# Count word frequencies
word_counts = Counter(all_words)

# Select the 10 most frequent words
top_10_words = word_counts.most_common(10)

# Display the results
print("Top 10 Most Common Words:")
for word, count in top_10_words:
    print(f"{word}: {count}")

# Prepare data for the chart
words = [item[0] for item in top_10_words]
counts = [item[1] for item in top_10_words]

# Plot the bar chart
plt.figure(figsize=(10, 5))
plt.bar(words, counts)
plt.title("Top 10 Most Common Words")
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

from sklearn.feature_extraction.text import TfidfVectorizer

# Create the TF-IDF vectorizer
vectorizer = TfidfVectorizer()

# Convert cleaned text into TF-IDF values
tfidf_matrix = vectorizer.fit_transform(df["cleaned_text"])

# Get the vocabulary
feature_names = vectorizer.get_feature_names_out()

# Calculate average TF-IDF score for each word across the dataset
average_scores = tfidf_matrix.mean(axis=0).A1

# Sort words by average TF-IDF score
sorted_indices = average_scores.argsort()[::-1][:10]

# Display top 10 keywords
print("Top 10 TF-IDF Keywords:")
for index in sorted_indices:
    print(f"{feature_names[index]}: {average_scores[index]:.4f}")
```
